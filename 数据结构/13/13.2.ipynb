```python
import tkinter as tk#引入tkinter画图

class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def build_tree(arr):
    if not arr:
        return None
    root = TreeNode(arr[0])
    queue = [root]
    i = 1
    while queue and i < len(arr):
        node = queue.pop(0)
        if arr[i] is not None:
            node.left = TreeNode(arr[i])
            queue.append(node.left)
        i += 1
        if i >= len(arr):
            break
        if arr[i] is not None:
            node.right = TreeNode(arr[i])
            queue.append(node.right)
        i += 1
    return root

class BinaryTreeDrawer:
    def __init__(self, root_node):
        self.root_node = root_node
        self.window = tk.Tk()
        self.window.title("二叉树绘制")
        self.canvas_width = 800
        self.canvas_height = 600
        self.canvas = tk.Canvas(self.window, width=self.canvas_width, height=self.canvas_height, bg="white")
        self.canvas.pack()

        self.node_radius = 20
        self.level_gap = 80
        self.draw_tree()

    def get_tree_depth(self, node):
        if not node:
            return 0
        return 1 + max(self.get_tree_depth(node.left), self.get_tree_depth(node.right))

    def draw_node(self, x, y, text):
        #画节点
        r = self.node_radius
        self.canvas.create_oval(x - r, y - r, x + r, y + r, fill="lightblue", outline="black")
        self.canvas.create_text(x, y, text=str(text), font=("Arial", 12, "bold"))

    def draw_edge(self, x1, y1, x2, y2):
        #画边
        self.canvas.create_line(x1, y1, x2, y2, fill="black")

    def draw_recursive(self, node, x, y, x_gap):
        #总绘制
        if not node:
            return
        self.draw_node(x, y, node.val)
        if node.left:
            left_x = x - x_gap
            left_y = y + self.level_gap
            self.draw_edge(x, y + self.node_radius, left_x, left_y - self.node_radius)
            self.draw_recursive(node.left, left_x, left_y, x_gap / 2)
        if node.right:
            right_x = x + x_gap
            right_y = y + self.level_gap
            self.draw_edge(x, y + self.node_radius, right_x, right_y - self.node_radius)
            self.draw_recursive(node.right, right_x, right_y, x_gap / 2)

    def draw_tree(self):
        depth = self.get_tree_depth(self.root_node)
        initial_gap = self.canvas_width / 4
        start_x = self.canvas_width / 2
        start_y = 50
        self.draw_recursive(self.root_node, start_x, start_y, initial_gap)

    def run(self):
        self.window.mainloop()

if __name__ == "__main__":
    arr = [10, 5, 15, 3, 7, None, 20]
    root = build_tree(arr)
    drawer = BinaryTreeDrawer(root)
    drawer.run()
```
