```python
import tkinter as tk#画二叉树用

class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    def insert(self, key):
        if self.root is None:
            self.root = Node(key)
        else:
            self._insert(self.root, key)

    def _insert(self, node, key):
        if key < node.key:
            if node.left is None:
                node.left = Node(key)
            else:
                self._insert(node.left, key)
        else:
            if node.right is None:
                node.right = Node(key)
            else:
                self._insert(node.right, key)

    def delete_by_predecessor(self, key):
        self.root = self._delete_by_predecessor(self.root, key)

    def _delete_by_predecessor(self, node, key):
        if node is None:
            return None
        if key < node.key:
            node.left = self._delete_by_predecessor(node.left, key)
        elif key > node.key:
            node.right = self._delete_by_predecessor(node.right, key)
        else:
            if node.left is None:
                return node.right
            elif node.right is None:
                return node.left
            pred = self._find_max(node.left)
            node.key = pred.key
            node.left = self._delete_by_predecessor(node.left, pred.key)
        return node

    def delete_by_successor(self, key):
        self.root = self._delete_by_successor(self.root, key)

    def _delete_by_successor(self, node, key):
        if node is None:
            return None
        if key < node.key:
            node.left = self._delete_by_successor(node.left, key)
        elif key > node.key:
            node.right = self._delete_by_successor(node.right, key)
        else:
            if node.left is None:
                return node.right
            elif node.right is None:
                return node.left
            succ = self._find_min(node.right)
            node.key = succ.key
            node.right = self._delete_by_successor(node.right, succ.key)
        return node

    def _find_max(self, node):
        current = node
        while current.right is not None:
            current = current.right
        return current

    def _find_min(self, node):
        current = node
        while current.left is not None:
            current = current.left
        return current

def draw_tree(bst, title):
    # 新建窗口画出二叉树，用的tkinter
    win = tk.Tk()
    win.title(title)
    canvas = tk.Canvas(win, width=800, height=500, bg='white')
    canvas.pack(fill=tk.BOTH, expand=True)

    def draw_node(node, x, y, dx):
        if node is None:
            return
        r = 20
        canvas.create_oval(x - r, y - r, x + r, y + r, fill='lightblue')
        canvas.create_text(x, y, text=str(node.key), font=('Times New Roman', 12, 'bold'))
        if node.left:
            lx, ly = x - dx, y + 60
            canvas.create_line(x, y + r, lx, ly - r)
            draw_node(node.left, lx, ly, dx // 2)
        if node.right:
            rx, ry = x + dx, y + 60
            canvas.create_line(x, y + r, rx, ry - r)
            draw_node(node.right, rx, ry, dx // 2)

    draw_node(bst.root, 400, 50, 200)
    win.mainloop()

if __name__ == "__main__":
    keys = [50, 30, 70, 20, 40, 60, 80]
    bst1 = BST()
    for k in keys:
        bst1.insert(k)
    draw_tree(bst1, "初始 BST")

    bst_pre = BST()
    for k in keys:
        bst_pre.insert(k)
    bst_pre.delete_by_predecessor(50)
    draw_tree(bst_pre, "删除根节点（中序前驱策略）")

    bst_succ = BST()
    for k in keys:
        bst_succ.insert(k)
    bst_succ.delete_by_successor(50)
    draw_tree(bst_succ, "题2：删除根节点（中序后继策略）")
```
