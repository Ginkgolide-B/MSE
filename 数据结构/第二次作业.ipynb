```python
from typing import Any
from collections.abc import Sequence

class OrderedArrayInsert:
    def __init__(self, iterable: Sequence[Any]):
        self.iterable = list(iterable)   '''创建副本，避免修改原序列'''

    def insert(self, element: Any) -> None:
        insert_index = len(self.iterable)
        for i, val in enumerate(self.iterable):
            if element < val:
                insert_index = i
                break

        self.iterable.insert(insert_index, element)

base_array = [1, 4, 21, 59]
ordered = OrderedArrayInsert(base_array)
ordered.insert(19)   # [1,4,19,21,59]
ordered.insert(5)    # [1,4,5,19,21,59]
ordered.insert(100)  # [1,4,5,19,21,59,100]
print(ordered.iterable)
```
