```python
import heapq

class MedianFinder:
    def __init__(self):
        # 大顶堆（存较小的一半），用负数模拟
        self.max_heap = []   # 元素取负，堆顶是最大值的负数
        # 小顶堆（存较大的一半）
        self.min_heap = []   # 堆顶是最小值

    def addNum(self, num: int) -> None:
        # 先默认加入大顶堆（较小的一半）
        heapq.heappush(self.max_heap, -num)
        # 把大顶堆的最大值（实际值）移到小顶堆，保证小顶堆的所有元素 >= 大顶堆的所有元素
        heapq.heappush(self.min_heap, -heapq.heappop(self.max_heap))
        
        # 平衡两堆大小：如果小顶堆比大顶堆多1个以上，则移回
        if len(self.min_heap) > len(self.max_heap) + 1:
            heapq.heappush(self.max_heap, -heapq.heappop(self.min_heap))
        # 或者当大顶堆比小顶堆大时（理论上不会，但为保险也可调整）
        # 通常我们保持 len(max_heap) >= len(min_heap) 或相差不超过1，这里采用更标准的：
        # 让大顶堆大小 >= 小顶堆大小（差不超过1）
        if len(self.max_heap) < len(self.min_heap):
            # 如果小顶堆更大，移一个到大顶堆
            heapq.heappush(self.max_heap, -heapq.heappop(self.min_heap))

    def findMedian(self) -> float:
        # 如果总数为奇数，中位数在大顶堆堆顶（因为大顶堆多一个元素或相等）
        if len(self.max_heap) > len(self.min_heap):
            return -self.max_heap[0]
        else:
            # 偶数，取两堆顶的平均值
            return (-self.max_heap[0] + self.min_heap[0]) / 2.0


# 测试用例（按题目示例）
mf = MedianFinder()
mf.addNum(3)
print(mf.findMedian())  # 3
mf.addNum(1)
print(mf.findMedian())  # 2
mf.addNum(4)
print(mf.findMedian())  # 3
mf.addNum(1)
print(mf.findMedian())  # 2
mf.addNum(5)
print(mf.findMedian())  # 3
```
