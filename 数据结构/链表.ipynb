```python
class Node:
    """链表结点类。"""
    def __init__(self, data: str | int | float):
        self.data: str | int | float = data
        self.next_node: Node | None = None

    def __repr__(self) -> str:
        return f"Node({self.data!r})"


class LinkedList:
    """链表类，支持删除节点操作。"""

    def __init__(self):
        self.first_node: Node | None = None

    def __repr__(self) -> str:
        """返回链表的字符串表示。"""
        if not self.first_node:
            return "LinkedList([])"
        nodes: list[str] = []
        current = self.first_node
        while current:
            nodes.append(str(current.data))
            current = current.next_node
        return f"LinkedList([{', '.join(nodes)}])"

    # ---------- 新增删除方法 ----------

    def remove(self, data: str | int | float) -> bool:
        """
        删除链表中第一个值为 data 的节点。
        :param data: 要删除的数据
        :return: 是否成功删除（True 表示找到并删除，False 表示未找到）
        """
        current = self.first_node
        prev = None

        while current:
            if current.data == data:
                if prev is None:
                    # 删除的是头节点
                    self.first_node = current.next_node
                else:
                    # 删除中间或尾节点
                    prev.next_node = current.next_node
                return True
            prev = current
            current = current.next_node
        return False

    def remove_at(self, index: int) -> bool:
        """
        删除链表中指定索引（从 0 开始）的节点。
        :param index: 要删除的位置
        :return: 是否成功删除（索引有效返回 True，否则 False）
        """
        if index < 0 or self.first_node is None:
            return False

        if index == 0:
            # 删除头节点
            self.first_node = self.first_node.next_node
            return True

        current = self.first_node
        prev = None
        pos = 0

        while current and pos < index:
            prev = current
            current = current.next_node
            pos += 1

        if current is None:   # 索引超出链表长度
            return False

        # 此时 current 指向待删除节点，prev 是其前驱
        prev.next_node = current.next_node
        return True

    def pop(self) -> str | int | float | None:
        """
        删除并返回链表的最后一个节点（尾节点）。
        :return: 尾节点的数据，若链表为空则返回 None
        """
        if self.first_node is None:
            return None

        # 只有一个节点
        if self.first_node.next_node is None:
            data = self.first_node.data
            self.first_node = None
            return data

        # 至少两个节点
        current = self.first_node
        while current.next_node and current.next_node.next_node:
            current = current.next_node
        # 此时 current 是倒数第二个节点
        tail = current.next_node
        data = tail.data
        current.next_node = None
        return data

    def clear(self) -> None:
        """清空整个链表（释放所有节点）。"""
        self.first_node = None

# 创建链表并添加一些节点（手动添加，或先实现 append 方法，这里略）
ll = LinkedList()
# 假设已有节点：10 -> 20 -> 30 -> 20 -> 40
# 为了演示，我们直接构造（实际可用 append）
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)
n4 = Node(20)
n5 = Node(40)
n1.next_node = n2
n2.next_node = n3
n3.next_node = n4
n4.next_node = n5
ll.first_node = n1
print(ll)          # LinkedList([10, 20, 30, 20, 40])

# 按值删除第一个 20
ll.remove(20)
print(ll)          # LinkedList([10, 30, 20, 40])

# 按索引删除位置 2（从0开始，即第三个元素 20）
ll.remove_at(2)
print(ll)          # LinkedList([10, 30, 40])

# 弹出尾节点
last = ll.pop()
print(last)        # 40
print(ll)          # LinkedList([10, 30])

# 清空链表
ll.clear()
print(ll)          # LinkedList([])
```
