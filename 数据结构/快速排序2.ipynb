```python
from typing import Sequence
import random

class QuickSelect:
    """
    快速选择类（三路划分版本），用于在无序数组中查找第k小的元素。
    平均时间复杂度：O(N)
    """

    def __init__(self, arr: Sequence[int | float]):
        """
        初始化快速选择器。

        :param arr: 要查询的数组
        """
        self.arr: list[int | float] = list(arr)
        self.length: int = len(arr)

    def partition_three_way(self, left_index: int, right_index: int) -> tuple[int, int]:
        """
        三路划分（荷兰国旗问题）。
        将 arr[left_index..right_index] 分为三部分：
            - 小于 pivot 的元素在 [left_index, lt-1]
            - 等于 pivot 的元素在 [lt, gt]
            - 大于 pivot 的元素在 [gt+1, right_index]

        :param left_index: 左边界索引
        :param right_index: 右边界索引
        :return: (lt, gt) 等于区间的左右边界
        """
        # 随机选择 pivot（从当前子数组中选取）
        pivot_idx = random.randint(left_index, right_index)
        pivot = self.arr[pivot_idx]

        # 将 pivot 交换到最左边，便于三路划分
        self.arr[left_index], self.arr[pivot_idx] = self.arr[pivot_idx], self.arr[left_index]

        lt = left_index      # 小于区的右边界
        gt = right_index     # 大于区的左边界
        i = left_index + 1   # 当前扫描指针

        while i <= gt:
            if self.arr[i] < pivot:
                # 当前元素小于 pivot，与 lt+1 交换，扩展小于区
                self.arr[lt], self.arr[i] = self.arr[i], self.arr[lt]
                lt += 1
                i += 1
            elif self.arr[i] > pivot:
                # 当前元素大于 pivot，与 gt 交换，扩展大于区
                self.arr[i], self.arr[gt] = self.arr[gt], self.arr[i]
                gt -= 1
            else:
                # 等于 pivot，指针后移
                i += 1

        # 最终 arr[lt..gt] 均为 pivot，左边小于，右边大于
        return lt, gt

    def select(
        self, k: int, left_index: int | None = None, right_index: int | None = None
    ) -> int | float:
        """
        查找第k小的元素（k从0开始计数）。

        :param k: 要找第k小的元素（0表示最小，length-1表示最大）
        :param left_index: 搜索范围的左边界
        :param right_index: 搜索范围的右边界
        :return: 第k小的元素的值
        """
        if left_index is None:
            left_index = 0
        if right_index is None:
            right_index = self.length - 1

        # 基准情形：只剩一个元素
        if left_index == right_index:
            return self.arr[left_index]

        # 三路划分，得到等于区间的左右边界
        lt, gt = self.partition_three_way(left_index, right_index)

        if k < lt:
            # 第k小在左子数组（小于区）
            return self.select(k, left_index, lt - 1)
        elif k > gt:
            # 第k小在右子数组（大于区）
            return self.select(k, gt + 1, right_index)
        else:
            # 第k小落在等于区，直接返回 pivot 值
            return self.arr[lt]


# ---------- 测试 ----------
if __name__ == "__main__":
    test_arr3: list[int] = [64, 34, 25, 12, 22, 11, 90]
    print(f"数组: {test_arr3}")
    print(f"排序后参考: {sorted(test_arr3)}")

    # 找第0小（最小）
    qs3: QuickSelect = QuickSelect(test_arr3.copy())
    print(f"\n第0小（最小）: {qs3.select(0)}")

    # 找第3小
    qs3 = QuickSelect(test_arr3.copy())
    print(f"第3小: {qs3.select(3)}")

    # 找第6小（最大）
    qs3 = QuickSelect(test_arr3.copy())
    print(f"第6小（最大）: {qs3.select(6)}")

    # 找中位数
    qs3 = QuickSelect(test_arr3.copy())
    median_index: int = len(test_arr3) // 2
    print(f"中位数（第{median_index}小）: {qs3.select(median_index)}")
```
